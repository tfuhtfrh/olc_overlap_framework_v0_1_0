## 当前最重要的术语对齐

目前对话中出现了三个容易混淆的“k”，今后统一区分：

### 1. Frontier 中的 $k_{\mathrm{trav}}$

旧 V4 / frontier 语境中的

$$
k_{\mathrm{trav}}
$$

表示一个局部 window 中同时穿过 interior 的 **traversal 条数**。

它不是 sampler move 的大小。

### 2. Exchange 中的 $k_{\mathrm{ex}}$

当前 sampler 实验中的

$$
k_{\mathrm{ex}}
$$

表示一次 exchange 中同时选择多少条已选边和多少条未选边：

$$
k_{\mathrm{ex}}\text{ 个 }1\to0
\quad + \quad
k_{\mathrm{ex}}\text{ 个 }0\to1.
$$

例如 $k_{\mathrm{ex}}=3$ 是一个 6-bit 联合翻转。

### 3. Reconnect

**reconnect 不是这里的 $k$。**

当前实现的 reconnect 是一个 directed 2-switch：

$$
(a\to b,\;c\to d)
\longleftrightarrow
(a\to d,\;c\to b),
$$

它是一个 4-bit move，同时保持总边数以及各顶点的入/出度。

以后若扩展到 3-edge reconnect，建议使用 $r_{\mathrm{rec}}$ 或直接写 “3-edge reconnect”，避免继续和 $k$ 混用。
